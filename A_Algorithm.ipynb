```python
graph={
    'S':[('A',4),('B',3)],
    'A':[('C',5),('D',2)],
    'B':[('E',6),('F',4)],
    'C':[('G',3)],
    'D':[('G',2)],
    'E':[('G',1)],
    'F':[('G',2)],
    'G':[]
}
#HEURISTIC VALUES

heuristic={
    'S':8,
    'A':5,
    'B':4,
    'C':3,
    'D':2,
    'E':1,
    'F':2,
    'G':0
}
def astar(start,goal):
  open_list=[(start,0)]
  came_from={}
  g_cost={start:0}

  while open_list:
    #select node with min f=g+h
    current=min(open_list, key=lambda x:x[1]+heuristic[x[0]])
    open_list.remove(current)
    current_node=current[0]

    if current_node == goal:
      path=[goal]

      while current_node in came_from:
        current_node=came_from[current_node]
        path.append(current_node)

      path.reverse()
      return path,g_cost[goal]

    # This loop should be at the same indentation level as the 'if current_node == goal:' block
    for neighbor,cost in graph[current_node]:
      new_cost=g_cost[current_node]+cost
      if neighbor not in g_cost or new_cost<g_cost[neighbor]:
        came_from[neighbor]=current_node
        g_cost[neighbor]=new_cost
        open_list.append((neighbor, new_cost))

  # This return should be outside the 'while open_list' loop
  return None, float('inf')



#Main Program
start=input("Enter the start node:")
goal=input("Enter the goal node:")

path,cost=astar(start,goal)

if path:
  print("\nShortest Path:","->".join(path))
  print("Total Cost:",cost)
else:
  print("Path not found")
```
